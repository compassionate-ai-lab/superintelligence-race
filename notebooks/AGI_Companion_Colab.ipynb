## 4. M1: the governance-capacity boundary (Chapter 1)

How many actions per day can a review team oversee at a target utilization? Change the inputs and re-run.

```python
import matplotlib.pyplot as plt
from agi_models.m1_governance import Oversight

N, r, T, k, mu = 10_000, 0.07, 8, 15, 6     # actions/day, review fraction, hours/day, reviewers, reviews per reviewer-hour
o = Oversight(N=N, r=r, T=T, k=k, mu=mu)
print(f"utilization rho = {o.utilization:.3f};  N_max at rho*=0.8: {o.n_max():,.0f} actions/day;  reviewers needed: {o.reviewers_needed()}")

rs = [x / 1000 for x in range(10, 121)]
plt.figure(figsize=(6, 3.2))
plt.plot([x * 100 for x in rs], [Oversight(N, x, T, k, mu).n_max() for x in rs])
plt.scatter([r * 100], [N], color="tab:orange", zorder=3, label="your system today")
plt.xlabel("fraction of actions needing review, r (%)"); plt.ylabel("N_max, actions per day")
plt.title("Oversight boundary (above the curve: delegation beyond effective oversight)", fontsize=9)
plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()
```

## 6. M5: minimax regret and the 8/11 threshold (Chapter 3)

```python
from agi_models import m5_regret as m5
L = m5.LOSS
ps = [i / 100 for i in range(101)]
plt.figure(figsize=(6, 3.2))
for a, v in L.items():
    plt.plot(ps, [p * v[0] + (1 - p) * (v[1] + v[2]) / 2 for p in ps], label=a.split(":")[0])
t = m5.threshold_fast(L, "S1 big bet now", "S4 staged: institutions now, prepared option to scale")
plt.axvline(t, ls="--", color="grey")
plt.text(t + .015, 14, f"S1 beats S4\nabove P(fast) = {t:.3f}\n(that is, 8/11)", fontsize=8, ha="left", va="center")
plt.xlabel("probability of the fast scenario"); plt.ylabel("expected loss"); plt.legend(fontsize=7)
plt.grid(alpha=.3); plt.tight_layout(); plt.show()
print("minimax-regret choice:", m5.minimax_regret(L)[0])
```
